## 2. Create a synthetic dataset

```python
# generate x and y
n = 20

np.random.seed(0)
x = -2 + 4 * np.random.rand(n)
x = np.sort(x)

beta_0 = 5.0
beta_1 = 1.5
sigma = 0.5

epsilon = sigma * np.random.normal(loc=0.0, scale=1, size=n)

y_theoretical = beta_0 + beta_1 * x
y = beta_0 + beta_1 * x + epsilon

# plot the data
fig, ax = plt.subplots()
ax.scatter(x, y, color='red', marker=marker_1, linewidth=line_width_1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Synthetic dataset with one predictor')
```

## 3. Simple Linear Regression (SLR)

Model:
$$
y = \beta_0 + \beta_1 x
$$

Solution:
$$
\begin{align}
\hat{\beta}_1 =& \frac{cov(X, Y)}{var(X)} \\
\hat{\beta}_0 =& \bar{y} - \hat{\beta}_1 \bar{x}
\end{align}
$$



```python
# --------------------------------------------------------------------------
# simple linear regression
# --------------------------------------------------------------------------
n = len(x)

x_bar = np.mean(x)
y_bar = np.mean(y)

# do linear regression using my own function
lm_d_result = d_lm_analytical_solution_class.d_lm(x, y)

# plot
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.scatter(x, y, color='red', marker=marker_1, linewidth=line_width_1)
ax.plot(x, y_theoretical, color='green', label='theoretical', linewidth=line_width_1)
ax.plot(x, lm_d_result['y_hat'], color='blue', label='predicted', linewidth=line_width_1)
ax.plot(x, np.ones(n)*y_bar, color='black', linestyle=':', linewidth=line_width_1)
ax.plot([x_bar, x_bar], [np.min(y), np.max(y)], color='black', linestyle=':', linewidth=line_width_1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title("Linear regression")
ax.legend(loc='lower right', fontsize=9)
# Other possible position of the legend: upper right, upper left, lower right, lower left, center left, center right, upper center, lower center
```

### 4. Multiple Linear Regression (MLR)

Model:
$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_0 + \cdots + \beta_d x_d
$$

Coefficients for MLR are computed using:

$$
\hat{\beta} = (X^T X)^{-1} X^T Y
$$

where the first column of X consists of 1 and the following columns are predictors and:
$$
\hat{\beta} = \begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \\ \vdots \\ \beta_d \end{bmatrix}
$$

```python
data_df = pd.read_csv('../data/exam_scores_for_linear_regression.csv')
print(f'\nDataset shape: {data_df.shape}')
print(f'\nFirst few rows of the dataset: \n{data_df.head()}')
```

```python
y = data_df[['exam_score']]
```

```python
# Simple linear regression: model y vs "study_hours" only

x1 = data_df[['study_hours']]
x1 = sm.add_constant(x1) # augment the ddata matrix by adding a column of 1`
print(x1.head())
```

```python
fig, ax = plt.subplots()
ax.scatter(data_df['study_hours'], data_df['exam_score'])
ax.set_xlabel('study hours')
ax.set_ylabel('exam score')
```

```python
simple_model = sm.OLS(y, x1).fit()
print(simple_model.summary())
```

```python
# Model y vs "study_hours" and 'gpa'

x2 = data_df[['study_hours', 'gpa']]
x2 = sm.add_constant(x2)
print(x2.head())
```

```python
multiple_model_1 = sm.OLS(y, x2).fit()
print(multiple_model_1.summary())
```

```python
# Model y vs "study_hours", 'gpa', and 'attendance'

x3 = data_df[['study_hours', 'gpa', 'attendance']]
x3 = sm.add_constant(x3)
print(x3.head())
```

```python
multiple_model_2 = sm.OLS(y, x3).fit()
print(multiple_model_2.summary())
```

## 5. Interpretation of results

- simple_model (study_hours only): study_hours explains 62.5% of the variance in exam scores ($R^2$ = 0.625). Each additional study hour is associated with a 5.34 point increase in exam score, and this effect is highly significant (p < 0.001).
- multiple_model_1 (study_hours + gpa): $R^2$ = 0.748 (adjusted 0.735), a meaningful improvement over simple_model. Both predictors are significant: study_hours' coefficient drops to 4.72 points per hour, and gpa adds 10.78 points per unit. The drop in the study_hours coefficient (5.34 → 4.72) shows that some of what simple_model attributed to study hours was actually confounded with gpa — students with higher GPAs tend to study more, so simple_model was partly picking up gpa's effect through study_hours.
- multiple_model_2 (study_hours + gpa + attendance): $R^2$ rises only to 0.762, and adjusted $R^2$ actually goes from 0.735 to 0.742 (a small gain). Attendance itself is not statistically significant (coefficient 0.176, p = 0.153), and adding it also inflates the condition number sharply (39.0 → 828), a sign of increasing multicollinearity among the predictors (attendance is likely correlated with study_hours and/or gpa). The intercept's p-value also crosses the 0.05 threshold (p = 0.050), reflecting the added noise from this extra, non-significant term.
- Taken together, multiple_model_1 looks like the best of the three: it captures nearly all the explanatory power gained by adding predictors to the simple_model, every coefficient is significant, and multiple_model_2's extra variable doesn't earn its added complexity — it neither significantly improves fit nor is itself a significant predictor, while making the model less numerically stable.
- study_hours and gpa are correlated, so including gpa changes (reduces) the estimated effect of study_hours, but attendance doesn't add a comparably confounded or independently useful signal here.

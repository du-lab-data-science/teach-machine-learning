### MLR: compute coefficients using:
$$
\hat{\beta} = (X^T X)^{-1} X^T Y
$$

```python
data_df = pd.read_csv('../data/exam_scores_for_linear_regression.csv')
print(data_df.shape)
print(data_df.head())
```

```python
y = data_df[['exam_score']]
```

```python
# Model y vs "study_hours" only

x1 = data_df[['study_hours']]
x1 = sm.add_constant(x1)
print(x1.head())
```

```python
fig, ax = plt.subplots()
ax.scatter(data_df['study_hours'], data_df['exam_score'])
ax.set_xlabel('study hours')
ax.set_ylabel('exam score')
```

```python
model_1 = sm.OLS(y, x1).fit()
print(model_1.summary())
```

```python
# Model y vs "study_hours" and 'gpa'

x2 = data_df[['study_hours', 'gpa']]
x2 = sm.add_constant(x2)
print(x2.head())
```

```python
model_2 = sm.OLS(y, x2).fit()
print(model_2.summary())
```

```python
# Model y vs "study_hours", 'gpa', and 'attendance'

x3 = data_df[['study_hours', 'gpa', 'attendance']]
x3 = sm.add_constant(x3)
print(x3.head())
```

```python
model_3 = sm.OLS(y, x3).fit()
print(model_3.summary())
```

Interpretation of the results

- model_1 (study_hours only): study_hours explains 62.5% of the variance in exam scores ($R^2$ = 0.625). Each additional study hour is associated with a 5.34 point increase in exam score, and this effect is highly significant (p < 0.001).
- model_2 (study_hours + gpa): R² = 0.748 (adjusted 0.735), a meaningful improvement over model_1. Both predictors are significant: study_hours' coefficient drops to 4.72 points per hour, and gpa adds 10.78 points per unit. The drop in the study_hours coefficient (5.34 → 4.72) shows that some of what model_1 attributed to study hours was actually confounded with gpa — students with higher GPAs tend to study more, so model_1 was partly picking up gpa's effect through study_hours.
- model_3 (study_hours + gpa + attendance): R² rises only to 0.762, and adjusted R² actually goes from 0.735 to 0.742 (a small gain). Attendance itself is not statistically significant (coefficient 0.176, p = 0.153), and adding it also inflates the condition number sharply (39.0 → 828), a sign of increasing multicollinearity among the predictors (attendance is likely correlated with study_hours and/or gpa). The intercept's p-value also crosses the 0.05 threshold (p = 0.050), reflecting the added noise from this extra, non-significant term.
- Taken together, model_2 looks like the best of the three: it captures nearly all the explanatory power gained by adding predictors to model_1, every coefficient is significant, and model_3's extra variable doesn't earn its added complexity — it neither significantly improves fit nor is itself a significant predictor, while making the model less numerically stable.
- study_hours and gpa are correlated, so including gpa changes (reduces) the estimated effect of study_hours, but attendance doesn't add a comparably confounded or independently useful signal here.

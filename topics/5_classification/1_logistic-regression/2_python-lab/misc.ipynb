```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-3, 3, 200)
y = np.exp(x)

plt.plot(x, y, label=r"$y=e^x$", color="red")

# Draw x = 0 and y = 0 axes
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Exponential Function $y=e^x$")
plt.legend()
plt.grid(True)

out_file_name = 'exponential_function.png'
plt.savefig(out_file_name)
plt.show()
```

## A tiny omics-like dataset
We simulate 40 samples and 1,000 measured features. This is deliberately **p >> n**.

```python
rng = np.random.default_rng(6210)
n_samples, n_features = 40, 1000
X = rng.normal(size=(n_samples, n_features))
y = np.array([0]*20 + [1]*20)
# Add a small group signal to the first five features.
X[y == 1, :5] += 0.8
feature_names = [f'feature_{j:04d}' for j in range(n_features)]
sample_names = [f'sample_{i:03d}' for i in range(n_samples)]
omics = pd.DataFrame(X, index=sample_names, columns=feature_names)
print('Shape:', omics.shape)
print('p/n:', n_features/n_samples)
omics.iloc[:5, :8]
```

```python
summary = pd.DataFrame({
    'mean': omics.mean(), 'sd': omics.std(),
    'missing_n': omics.isna().sum()
})
summary.head(10)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(7,4))
sns.histplot(omics.var().values, bins=35, ax=ax)
ax.set(xlabel='Feature variance', title='Variance across 1,000 simulated features')
plt.show()
```
